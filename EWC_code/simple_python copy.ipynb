```python
import numpy as np
import matplotlib.pyplot as plt

def gaussian_matrix(n):
    return np.random.normal(0, 1, (n, n))

def symmetric_gaussian_matrix(n):
    G = gaussian_matrix(n)
    return (G + G.T) / np.sqrt(2)

def spike(n, k):
    v = np.random.normal(0, 1, n)
    return k * np.outer(v, v)

def spiked_gaussian_matrix(n, k):
    G = symmetric_gaussian_matrix(n)/np.sqrt(n)
    S = spike(n, k)
    return G + S
```

```python
for n in [100, 500, 1000]:

    M = spiked_gaussian_matrix(n, k=0)

    eigenvalues = np.linalg.eigvals(M)
    plt.hist(eigenvalues, bins=50, density=True)
plt.show()
```
